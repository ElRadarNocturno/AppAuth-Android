# Gemini API: Getting started with Interactions API


The [Interactions API](https://ai.google.dev/gemini-api/docs/interactions?ua=chat) is a unified interface for building with Gemini models and agents. It simplifies the development of agentic applications by handling server-side state management, tool orchestration, and long-running tasks.

With a single endpoint, you can:

- Interact with Gemini models for text, image, and audio generation
- Build multi-turn conversations without managing history client-side
- Call custom functions and built-in tools like Google Search
- Run specialized agents like Deep Research for complex tasks

Once the installation is complete, you can start using the Gemini API. For a full quickstart, it's highly recommended to use the `Get_started.ipynb` notebook linked above, as it contains the most up-to-date guidance and examples for the Interactions API.

```
import google.generativeai as genai
from google.colab import userdata

# Recupera tu API key de los secretos de Colab
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')

# Configura el SDK de Gemini con tu API key
genai.configure(api_key=GOOGLE_API_KEY)

print('SDK de Gemini configurado con éxito.')
```

```
import google.generativeai as genai

# Lista los modelos disponibles para verificar que la API key funciona
print('Listando modelos disponibles:')
for m in genai.list_models():
  if 'generateContent' in m.supported_generation_methods:
    print(m.name)
```

Una vez que hayas ejecutado el código anterior, tu SDK de Gemini estará configurado y listo para ser utilizado. Puedes probarlo inicializando un modelo como el siguiente:

```
model = genai.GenerativeModel('gemini-pro-latest')
print("Modelo Gemini inicializado y listo.")
```

```
prompt = "Escribe una breve historia sobre un viaje al espacio."
response = model.generate_content(prompt)
print(response.text)
```
